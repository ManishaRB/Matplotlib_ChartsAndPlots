# Bar Charts

```python
import matplotlib.pyplot as plt
import numpy as np 
import pandas as pd 
```

```python
# Assigning x and y coordinates
CoursesName = ['C Course', 'C++ Course', 'Java Course', 'Python Course']
Student = [90, 60, 130, 190]

plt.bar(CoursesName, Student, color='green' ,width = 0.3, zorder=2)
plt.grid(True, axis='y', zorder=0)
plt.xlabel('Courses Name')
plt.ylabel('Student Registrations number')
plt.title('Number of Students in Different Courses')

plt.show()
```

```python
#Note: For horizontal bars, use height instead of width.
plt.barh(CoursesName,Student,  color='green',height = 0.3)

plt.xlabel('Student Registrations number')
plt.ylabel('Courses Name')


# Displaying the title
plt.title('Number of Students in Different Courses')

plt.show()
```

```python
male = [90, 60, 130, 600]
female = [100, 30, 160, 690]

plt.bar(CoursesName, male, width=0.5, label='male')
plt.bar(CoursesName, female, bottom=male, width=0.5, label='Female')
plt.title('Number of Students Per Course')
plt.xlabel('Course Name')
plt.ylabel('Number of students')
plt.legend()
plt.show()
```

```python
years = ['2016', '2017', '2018', '2019']
rice_price = [57, 50, 47, 30]
wheat_price = [43, 50, 53, 70]
```

```python
rice =  np.array([57, 50, 47, 30])
wheat =  np.array([43, 50, 53, 70])
mustard =  np.array([10, 16, 13, 15])

plt.figure(figsize=(10, 6))

plt.bar(years, rice, label='Rice')
plt.bar(years, wheat, bottom=rice, label='Wheat')
plt.bar(years, mustard, bottom=rice+wheat, label='Mustard seeds')

plt.xlabel('Years', fontsize=14)
plt.ylabel('Preference (%)', fontsize=14)
plt.title('The results of Rice, Wheat & Mustard seeds export', fontsize=20)
plt.legend()
plt.show()
```

```python
CoursesName = ['C Course', 'C++ Course', 'Java Course', 'Python Course']
StudentGroup1 = [90, 60, 130, 600]
StudentGroup2 = [100, 30, 160, 690]
```

```python
width = 0.3

# Calculate the position for the bars
x = np.arange(len(CoursesName))

# Plotting
plt.bar(x - width/2, StudentGroup1, color='red', width=width, label='Student Group 1')
plt.bar(x + width/2, StudentGroup2, color='green', width=width, label='Student Group 2')

plt.xlabel('Courses Name')
plt.ylabel('Number of Students')

plt.title('Number of Students in Different Courses')

plt.xticks(x, CoursesName, rotation=45, ha='right')  # Rotate x-axis labels for better readability

plt.legend()  # Add legend
plt.show()
```

```python
# Sample data
data = {'CoursesName': ['C Programming Course', 'C++ Programming Course', 'Java Programming Course', 'Python Programming Course'],
        'Student': [90, 60, 130, 600]}

# Create DataFrame
df = pd.DataFrame(data)

# Plotting
plt.bar(df['CoursesName'], df['Student'], color='green', width=0.45)

plt.xlabel('Courses Name')
plt.ylabel('Number of Students')

plt.title('Number of Students in Different Courses')

plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for better readability

plt.show()
```

```python
data = {'CoursesName': ['C Programming Course', 'C++ Programming Course', 'Java Programming Course', 'Python Programming Course'],
        'Student Group 1': [90, 60, 130, 600],
        'Student Group 2': [100, 30, 160, 690]}

df = pd.DataFrame(data)

width = 0.35
x = df.index

plt.bar(x - width/2, df['Student Group 1'], width=width, label='Student Group 1')
plt.bar(x + width/2, df['Student Group 2'], width=width, label='Student Group 2')
plt.xticks(x, df['CoursesName'], rotation=45)
plt.title('NUmber of Students Per Course')
plt.ylabel('Number of Students')
plt.xlabel('Course Name')
plt.legend()
plt.show()
```
