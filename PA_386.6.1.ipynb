### PA - 386.6.1 - Analyzing Monthly Expenses

```python
import matplotlib.pyplot as plt
import numpy as np 
import pandas as pd 
```

```python
expense_categories = ['Groceries', 'Utilities', 'Transportation', 'Dining Out',  'Entertainment']
amount_spent = [500, 300, 200, 400, 250]

# Create horizontal bar chart
plt.barh(expense_categories, amount_spent, color=['blue', 'orange', 'green', 'red', 'purple'])

# Labels and title
plt.xlabel('Amount Spent')
plt.ylabel('Expense Categories')
plt.title('Monthly Expenses Distribution')

# Display chart
plt.show()
```
